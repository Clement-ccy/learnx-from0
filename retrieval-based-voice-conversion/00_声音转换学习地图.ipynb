# 00 · 一句话换个声音来讲

同一句“今天出门走走吧”，由不同的人说，内容相同，听感却很不同。声音转换想研究的就是：**保留说了什么和说话的节奏，同时改变音色。**

这是一页学习计划。我们会从自有或已授权的短录音出发，先看见声音的结构，再理解内容特征、音高和检索怎样参与转换。

把它想成一段演奏：内容像乐句，F0 勾勒高低走向，音色让你辨认是哪种乐器。接下来把这几条线索逐一拆开听。

## 1. 先把声音画出来

**波形**沿时间记录振幅，适合看停顿、起音和强弱。一段 16 kHz 录音，每秒保存 16000 个采样值。

**STFT** 像拿一个短窗口沿着声音移动，逐段看里面有哪些频率。窗口长一些，频率看得更细；窗口短一些，时间变化更清楚。**梅尔频谱**再按感知尺度汇聚频带，让频率结构更便于观察。

**F0** 则追踪有声段的基频，比如一句话的音调怎样升降。气声、摩擦音等片段要结合清浊音信息来看。

第一组计划实验：同一段录音，同时看波形、频谱、梅尔频谱和 F0 轨迹，找到一句话里停顿、元音和辅音各自留下的痕迹。

## 2. 内容、音高、音色，各走哪条路？

内容特征像一份发音草稿，尽量留下说话内容；F0 带着音高走向；检索则从目标声音的特征库里寻找相近片段，给转换提供参考。

具体组件会按选定的 RVC 版本展开。学习时先跟踪一段录音：特征怎样提取，检索结果怎样混合，最后又怎样交给生成器。

第二组计划实验：固定输入，分别改变音高处理和检索比例，听同一句话在哪些地方发生了变化。

## 3. 回到耳朵：输出听起来怎么样？

特征经过转换与合成，最后又成为可以播放的波形。把输入和输出交替听，分别关注发音是否清楚、停顿与节奏怎样变化、音色有什么特征，以及哪里出现了金属感或断裂。

后续选定实现版本后，会建立独立 uv 环境，把这些计划写成“听一段 → 看一张图 → 改一个参数 → 再听”的实验。
